```python
def parsear_semana(week_url, semana):

    html = requests.get(week_url).text

    soup = BeautifulSoup(html, "lxml")

    texto = soup.get_text("\n")

    lineas = [x.strip() for x in texto.split("\n") if x.strip()]

    texto_plano = " ".join(lineas)

    # ==================================================
    # LECTURA SEMANAL
    # ==================================================

    m = re.search(
        r"([A-ZÁÉÍÓÚÑ]+(?:\s+[A-ZÁÉÍÓÚÑ]+)*\s+\d+,\s*\d+)",
        texto_plano
    )

    lectura_semanal = m.group(1) if m else None

    # ==================================================
    # CANCIONES
    # ==================================================

    canciones = re.findall(r"Canci[oó]n\s+(\d+)", texto_plano)

    canciones = [int(x) for x in canciones]

    cancion_inicio = canciones[0] if len(canciones) > 0 else None
    cancion_intermedia = canciones[1] if len(canciones) > 1 else None
    cancion_final = canciones[2] if len(canciones) > 2 else None

    # ==================================================
    # INTRODUCCIÓN
    # ==================================================

    m = re.search(
        r"(Palabras de introducción\s*\(\d+\s*min[s]?\.\))",
        texto_plano
    )

    introduccion = m.group(1) if m else None

    # ==================================================
    # CONCLUSIÓN
    # ==================================================

    m = re.search(
        r"(Palabras de conclusión\s*\(\d+\s*min[s]?\.\))",
        texto_plano
    )

    conclusion = m.group(1) if m else None

    # ==================================================
    # PARSER ITEMS
    # ==================================================

    SECCIONES = [
        "TESOROS DE LA BIBLIA",
        "SEAMOS MEJORES MAESTROS",
        "NUESTRA VIDA CRISTIANA"
    ]

    registros = []

    seccion_actual = None
    item_actual = None

    for linea in lineas:

        # ----------------------------------------
        # detectar sección
        # ----------------------------------------

        if linea in SECCIONES:

            seccion_actual = linea
            continue

        # ----------------------------------------
        # detectar item
        # ----------------------------------------

        m_item = re.match(r"^(\d+)\.\s(.+)", linea)

        if m_item:

            numero = int(m_item.group(1))
            titulo = m_item.group(2)

            item_actual = {
                "semana": semana,
                "lectura_semanal": lectura_semanal,

                "seccion": seccion_actual,
                "item_num": numero,
                "titulo": titulo,

                "minutos": None,

                "tipo_actividad": None,
                "referencia": None,
                "descripcion": None,

                "cancion_inicio": cancion_inicio,
                "cancion_intermedia": cancion_intermedia,
                "cancion_final": cancion_final,

                "introduccion": introduccion,
                "conclusion": conclusion,

                "url": week_url
            }

            registros.append(item_actual)

            continue

        # ----------------------------------------
        # detectar duración
        # ----------------------------------------

        m_min = re.search(r"\((\d+)\s*min", linea)

        if m_min and item_actual is not None:

            item_actual["minutos"] = int(m_min.group(1))

            item_actual["descripcion"] = linea

            # ----------------------------------------
            # tipo actividad
            # ----------------------------------------

            tipos = [
                "DE CASA EN CASA",
                "PREDICACIÓN INFORMAL",
                "CURSO BÍBLICO"
            ]

            for t in tipos:

                if t in linea:
                    item_actual["tipo_actividad"] = t

            # ----------------------------------------
            # referencias
            # ----------------------------------------

            refs = re.findall(r"\(([A-Za-z].*?)\)", linea)

            if refs:
                item_actual["referencia"] = " | ".join(refs)

    return pd.DataFrame(registros)
```
